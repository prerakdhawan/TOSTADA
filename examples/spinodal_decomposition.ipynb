# Spinodal Decomposition in 2D/3D

***

#### The physics module of `tostada` contains a Cahn Hilliard solver to describe phase separation dynamics in fluids.

$\Large \frac{dc}{dt} = \triangledown \cdot \left(Dc \triangledown \left( f'(c) - \gamma \triangledown^2 c\right) \right)$

#### Using different initial conditions, different phase distributions can be obtained. In this notebook, we shall create a spinodal decomposed media available in `tostada.Phaseprocess` and later explore its statistical properties using the constructed `tostada.PhaseDistribution` object.

```python
import numpy as np
import matplotlib.pyplot as plt
import tostada as tost
```

```python
process = tost.Phaseprocess(volumefraction=0.5,
                            BoxSize=7,
                            resolution=0.02,
                            interpore_distance=0.2,
                            is_3D = False)
```

#### To obtain a spinodally decomposed media, one can use the `spinodal_decomposed_microstructure()` method in `Phaseprocess`. It essentially uses the `Cahn-Hilliard` solver available in `tostada.physics` module. Since our process is defined in 2D, the resulting phase distribution will also be in 2D.

```python
phase = process.spinodal_decomposed_microstructure(D=20, #Diffusion constant
                                                   gamma=0.3, # interface parameter deciding width of transition region
                                                   dt=0.002, # time-stepping 
                                                   num_iter=4000, # nuSmber of iterations to run
                                                   frame_iter=100, # frequency at which the 2D/3D slices are saved
                                                   animate=False # animation of phase distribution
                                                   ) 
```

#### The phase evolution keeps track of the mean center-to-center distance of phases (surface-surface or void-void correlation length) throughout the time-stepping. Since the Diffusion coefficient is constant at all times, the phase separation will not arrest by itself and therefore once the system reaches the “scaling regime” (meaning the system starts showing self-similarity at any given time), the simulation can be terminated by tuning the `num_iter` option as and when the desired correlation length is reached.

#### The histogram of phases can quantitatively tell the occurance of intermediate phase values.

```python
plt.hist(phase.image)
```

#### As it can be seen, most of the pixels are either 0s or 1s with only a narrow range representing the interface. The width of the interface region can be increased by increasing the `gamma` parameter and proportionately decreasing `D`.

```python
phase = process.spinodal_decomposed_microstructure(D=3, #Diffusion constant
                                                   gamma=2, # interface parameter deciding width of transition region
                                                   dt=0.002, # time-stepping 
                                                   num_iter=4000, # nuSmber of iterations to run
                                                   frame_iter=100, # frequency at which the 2D/3D slices are saved
                                                   animate=False # animation of phase distribution
                                                   ) 
```

```python
plt.hist(phase.image)
```

#### Clearly, histogram shows a broader interface region as many more pixels are between 0s and 1s. The reciprocal space can further reveal the statistics. This is available via `tost.Visualize`

```python
Fig = tost.Visualize(phase)
Fig.plot_distribution(cmap='RdBu')
Fig.plot_reciprocal_space()
```

#### One can also plot the auto-correlation function (ACF) for this distribution to understand real-space correlations along the principle axes.

```python
Fig.plot_principle_ACF(ax=None)
```

#### Based on above data, the spectral density $\tilde{X}(q) \rightarrow 0$ for $q\rightarrow 0$ and the real-space correlations decaying at large lengths, this hints at hyperuniformity in the system. This can easily be evaluated from the methods of `tostada.PhaseDistribution` using `self.Hyperuniformity_data()` that gives the full-width at half-maximum of the spectral peak and the hyperuniformity index H of the distribution.

```python
phase.Hyperuniformity_data()
```

#### Based on these values, the system here can loosely be classified as nearly-hyperuniform. This system can possess a higher degree of hyperuniformity by tuning the phase decomposition parameters in the initialization and increasing the spatial size of the simulation.

#### Also, please note that similar system can be obtained in 3D by setting `is_3D = True` in the process initialization. For the sake of quicker simulations, we skip this here. Since 3D simulations scale are memory intensive, presence of a GPU (automatically detected in tostada) will greately help.
