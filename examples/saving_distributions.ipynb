## Saving and Loading distributions

***

The distributions created from `tostada.PointDistribution` or `tostada.PhaseDistribution` can conveniently be saved and loaded using `save()` and `load()` methods available in both distributions, respectively.

```python
import numpy as np
import tostada as tost
```

```python
process = tost.Pointprocess(10,0.3,0.5)
points=process.rect_lattice(noise=0.5,correlation_length=2)
```

### This distribution can be saved using `PointDistribution.save()` method with a given filename and directory path.

```python
points.save('rect_lattice')
```

#### The same file can easily be loaded without declaring the PointDistribution class. This is done through a staticmethod `load()` function.

```python
new_distribution = tost.PointDistribution.load('rect_lattice')
```

To check if everything is loaded correctly, let us call the plotting tool using `tostada.plot_util.Visualize`.

```python
fig = tost.Visualize(new_distribution)
fig.plot_distribution()
fig.plot_reciprocal_space(vmax=10)
```

```python
phase = new_distribution.Phaseobject(dx=0.01)
```

```python
fig = tost.Visualize(phase)
fig.plot_distribution()
```

Similar to `PointDistribution`, one can also use `save()` and `load()` methods for a `PhaseDistribution`. 

```python
phase.save('rect_lattice_phase')
```

One thing to note while using `save` and `load` methods is that if reciprocal space and real space quantities are computed, saving the entire state can be memory intensive, especially when dealing with 3D datasets (>200 MB). A future release will have an additional optional for a minimilist save.
